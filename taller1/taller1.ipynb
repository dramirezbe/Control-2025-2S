```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parámetros dados ---
wn = 3  # Frecuencia natural no amortiguada (radio)
theta_deg = 28.68  # Ángulo en grados
sigma = 0.2  # Factor de decaimiento (parte real)

# --- Conversión y cálculos ---
# Convertir el ángulo a radianes
theta_rad = np.deg2rad(theta_deg)

# --- Crear el gráfico ---
fig, ax = plt.subplots(figsize=(8, 8))

# --- 1. Trazar el círculo para wn ---
# Generar ángulos para un semicírculo en el semiplano izquierdo
angles = np.linspace(np.pi / 2, 3 * np.pi / 2, 200)
x_circle = wn * np.cos(angles)
y_circle = wn * np.sin(angles)
ax.plot(x_circle, y_circle, 'r--', label=f'$w_n = {wn}$')

# --- 2. Trazar la línea para theta (coeficiente de amortiguamiento) ---
# La línea se traza desde el origen con el ángulo theta respecto al eje real negativo
x_line = np.array([0, -wn * np.cos(theta_rad)])
y_line = np.array([0, wn * np.sin(theta_rad)])
ax.plot(x_line, y_line, 'g-', label=f'$\\theta = {theta_deg}°$ (línea de amortiguamiento)')
# También se puede trazar la línea simétrica en el cuadrante inferior izquierdo
y_line_neg = np.array([0, -wn * np.sin(theta_rad)])
ax.plot(x_line, y_line_neg, 'g-')


# --- 3. Trazar la línea vertical para sigma ---
ax.axvline(x=-sigma, color='b', linestyle='-.', label=f'$\\sigma = {sigma}$ (parte real)')

# --- Configuración del gráfico ---
ax.set_xlabel('Eje Real ($\\sigma$)')
ax.set_ylabel('Eje Imaginario ($j\\omega$)')
ax.set_title('Región en el Plano Complejo')
ax.grid(True, which='both')
ax.axhline(y=0, color='k')
ax.axvline(x=0, color='k')
ax.legend()

# Establecer límites para una mejor visualización
ax.set_xlim(-wn - 0.5, 1)
ax.set_ylim(-wn - 0.5, wn + 0.5)
ax.set_aspect('equal', adjustable='box')

# Mostrar el gráfico
plt.show()
```
