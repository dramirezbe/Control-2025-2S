```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import control as ctl

# Setup (reusing previous values)
step = 0.01
end_f = 10
f = np.linspace(step, end_f, int(end_f/step)) 

# Manual calculation using s = j*omega
omega = f
s = 1j * omega
```

# 6,7 A)

```python

# Define the transfer function G2
Numerator = (s+0.5)
Denominator = (s*(5*s+1)*(s**2+0.2*s+0.6))
G2 = Numerator / Denominator

# Get magnitude and phase
mag = np.abs(G2)
mag_db = 20 * np.log10(mag)
phase_deg = np.degrees(np.unwrap(np.angle(G2))) 

# Create 2 subplots (rows)
fig, ax = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Plot Magnitude (dB)
ax[0].semilogx(omega, mag_db)
ax[0].set_ylabel('Magnitude [dB]')
ax[0].grid(True, which="both")
# Add vertical line at the lowest plotted frequency (0.01 rad/s)
ax[0].axvline(step, color='r', linestyle='--')
ax[0].axvline(1, color='r', linestyle='--')

# Plot Phase (deg)
ax[1].semilogx(omega, phase_deg)
ax[1].set_ylabel('Phase [deg]')
ax[1].set_xlabel('Frequency [rad/s]')
ax[1].grid(True, which="both")
# Add vertical line at the lowest plotted frequency (0.01 rad/s)
ax[1].axvline(step, color='r', linestyle='--')
ax[1].axvline(1, color='r', linestyle='--')

plt.tight_layout()
plt.show()
```

# 6.21

```python
# System Definition
s = ctl.tf('s')

def plot_nyq(k=1):
    
    G = k / ((s+1)*(s**2+2*s+2))
    fig, ax = plt.subplots(1, 2, figsize=(14, 6))
    ctl.pzmap(G, ax=ax[0])
    ctl.nyquist(G, ax=ax[1])
    plt.show()

interact(plot_nyq, k=(-10, 10, 0.05))
```

# 6.26

```python
# Define G(s)
G = (25 * (s + 1)) / (s * (s + 2) * (s**2 + 2*s + 16))

# Calculate Margins
# gm = Gain Margin (ratio), pm = Phase Margin (deg)
# wg = Phase Crossover Freq, wp = Gain Crossover Freq
gm, pm, wg, wp = ctl.margin(G)
gm_db = 20 * np.log10(gm) # Convert ratio to dB

print(f"Gain Margin: {gm_db:.2f} dB")
print(f"Phase Margin: {pm:.2f} degrees")

# Generate Bode Plot with margins indicated
ctl.bode_plot(G, dB=True, display_margins=True)
plt.show()
```

```python
# Define G(s)
def plot_margins(K=1):
        G = (K*(s+2))/((s-1)*(s**2+2*s+2))

        # Calculate Margins
        # gm = Gain Margin (ratio), pm = Phase Margin (deg)
        # wg = Phase Crossover Freq, wp = Gain Crossover Freq
        gm, pm, wg, wp = ctl.margin(G)
        gm_db = 20 * np.log10(gm) # Convert ratio to dB

        print(f"Gain Margin: {gm_db:.2f} dB")
        print(f"Phase Margin: {pm:.2f} degrees")
        print(f"Phase Crossover Freq: {wp:.2f} rad/s")
        print(f"Gain Crossover Freq: {wg:.2f} rad/s")

        fig1, ax1 = plt.subplots(1, 2, figsize=(14, 6))
        fig2, ax2 = plt.subplots(1, 1, figsize=(7, 6))
        #ctl.root_locus(G, ax=ax[0])

        # Generate Bode Plot with margins indicated
        ctl.bode_plot(G, dB=True, display_margins=True, ax=ax1)
        ctl.root_locus(G, ax=ax2)
        plt.show()

interact(plot_margins, K=(0.1, 10, 0.1))
```
